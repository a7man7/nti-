```python
matrix = [[0 for _ in range(3)] for _ in range(3)]

for i in range(3):
    for j in range(3):
        if i == j:
            matrix[i][j] = 1
        if i + j == 2:
            matrix[i][j] = 2
        if i == 1 and j == 1:
            matrix[i][j] = 3

for row in matrix:
    print(" ".join(str(cell) for cell in row))
```
